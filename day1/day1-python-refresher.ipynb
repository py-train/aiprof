### 1.1 — Print and Comments

````python
# This is a comment — Python ignores it. Use comments to explain your code.

print("Hello, world!")          # print() displays text on screen
print("The training starts now.")
print(3 + 4)                    # you can print numbers too
````

> `print()` is how a program talks back to you. Comments (lines starting with `#`) are notes for humans — the computer skips them.

### 1.2 — Data Types

````python
name = "Claude"         # str  — text, always in quotes
year = 2025             # int  — whole number
score = 9.5             # float — decimal number
is_ready = True         # bool — True or False (capital T/F)

print(type(name))       # type() tells you what kind of value something is
print(type(year))
print(type(is_ready))
````

> Every value in Python has a type. Knowing the type tells you what you can do with it (you can do math with numbers; you can't divide two strings).

---

### 1.3 — Lists

````python
tools = ["ChatGPT", "Copilot", "Claude", "Gemini"]   # a list: ordered, can hold many items

print(tools[0])          # index 0 = first item
print(tools[-1])         # index -1 = last item
print(len(tools))        # len() counts the items

tools.append("Mistral")  # add an item to the end
print(tools)
````

> Lists are containers. Square brackets `[]` define them; each item is separated by a comma. Index numbers start at 0, not 1.

---

### 1.4 — Dictionaries

````python
model = {
    "name": "gpt-4o-mini",    # key: value pairs
    "provider": "OpenAI",
    "context_window": 128000
}

print(model["name"])          # access a value by its key
print(model["context_window"])

model["cost"] = "low"         # add a new key-value pair
print(model)
````

> Dictionaries store named data — like a form where each field has a label. You look up values by name (key), not by position.

### 1.5 — If / Elif / Else

````python
temperature = 38   # try changing this value and re-running

if temperature > 40:
    print("Too hot — shut down the server")
elif temperature > 35:
    print("Running warm — monitor closely")
else:
    print("Temperature is normal")
````

> Indentation (the spaces at the start of lines) is how Python groups code that belongs together. Everything indented under an `if` only runs if that condition is true.

**Try it:** Change `temperature` to `42`, then to `30`. Re-run each time.

### 1.6 — For Loops

````python
models = ["gpt-4o-mini", "claude-haiku", "gemini-flash"]

for model in models:                  # for each item in the list...
    print(f"Running test on: {model}")  # f-strings let you embed variables in text
````

> A `for` loop runs the same block of code once for each item in a list. The variable `model` takes a different value each iteration.

---

### 1.7 — While Loops

````python
count = 0

while count < 3:          # keep going as long as this is True
    print(f"Attempt {count + 1}")
    count += 1            # count = count + 1 (move toward the stopping condition)

print("Done")
````

> A `while` loop keeps running until its condition becomes False. Always make sure something changes inside the loop, or it will run forever.

---

### 1.8 — Functions

````python
def summarize(text, max_words):          # def defines a function; parameters are inputs
    words = text.split()                 # split text into a list of words
    short = words[:max_words]            # take the first max_words items
    return " ".join(short) + "..."       # join them back into a string and return it

result = summarize("Generative AI is transforming how we work with information", 5)
print(result)
````

> Functions are reusable blocks of code. `def` names them; `return` sends a value back to the caller. You call a function by name with arguments in parentheses.

### 1.9 — Built-in Functions

````python
sentence = "Generative AI is changing everything"

print(len(sentence))            # number of characters
print(len(sentence.split()))    # number of words
print(type(sentence))           # what type is it?

numbers = [4, 1, 7, 2, 9]
print(max(numbers))             # largest value
print(min(numbers))             # smallest value
print(sorted(numbers))          # sorted copy

print(range(5))                 # range(5) = 0, 1, 2, 3, 4
print(list(range(5)))           # convert to a list to see the values
````

## Section 2: Reading Real Code (10 min)

Now read a complete program — the number guessing game. You've seen all the pieces in Section 1; now see how they fit together.

````python
import random                          # bring in the random number toolkit

secret = random.randint(1, 20)         # pick a secret number between 1 and 20
attempts = 0                           # start a counter at zero

print("I'm thinking of a number between 1 and 20.")

while True:                            # keep looping until we break out
    guess = int(input("Your guess: ")) # ask the user to type a number; convert to int
    attempts += 1                      # add 1 to the counter

    if guess < secret:
        print("Too low!")
    elif guess > secret:
        print("Too high!")
    else:
        print(f"Correct! You got it in {attempts} attempts.")
        break                          # exit the loop — game over
````

## Section 3: Hello, LLM (15 min)

👀 **Everyone** | 💻 **Run each cell**

Now let's call an AI model with code. This is the same mechanism behind ChatGPT, Copilot, and every AI product.

### Step 3a — Read it first

````python
from openai import OpenAI             # import the official OpenAI library

client = OpenAI()                     # create a connection to the API
                                      # the API key is loaded from the environment automatically

response = client.chat.completions.create(
    model="gpt-4o-mini",              # which model to use
    messages=[                        # the conversation — a list of dicts (sound familiar?)
        {
            "role": "user",           # this message is from the user
            "content": "What are three habits of effective meetings?"
        }
    ]
)

print(response.choices[0].message.content)   # extract the model's reply and print it
````

**Five things to notice:**

| Line | What it is | Plain English |
|------|-----------|---------------|
| `from openai import OpenAI` | Import | "Load the OpenAI toolkit" |
| `client = OpenAI()` | Setup | "Open a connection to the API" |
| `model="gpt-4o-mini"` | Model selection | "Which AI brain to use" |
| `messages=[...]` | The prompt | "The conversation — a list of dicts" |
| `response.choices[0].message.content` | The reply | "The text the model wrote back" |

## Optional Extension: Multi-Turn Conversation (10 min)

💻 **For technical fast-finishers**

Our current code sends one message and gets one reply. Modify it to keep a running conversation:

````python
from openai import OpenAI

client = OpenAI()

conversation = [
    {"role": "system", "content": "You are a helpful professional assistant."}
]

print("Type 'quit' to end.\n")

while True:
    user_input = input("You: ")
    if user_input.lower() == "quit":
        break

    conversation.append({"role": "user", "content": user_input})

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=conversation        # send the full history every time
    )

    reply = response.choices[0].message.content
    conversation.append({"role": "assistant", "content": reply})  # save the reply too

    print(f"\nAssistant: {reply}\n")
````

**What's new:** The `conversation` list grows with every turn. The model "remembers" because we send the full history each time — not because it has actual memory.

Try it: start a conversation, then reference something you said two turns earlier.

> This is how ChatGPT works. The "memory" is just the message list.
